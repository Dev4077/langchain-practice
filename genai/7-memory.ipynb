### Agent with Memory


```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
```

```python
agent=create_agent(
    model="gpt-5",
    checkpointer=InMemorySaver()
)
```

```python
### Run with thread ID
config={"configurable": {"thread_id": "test1"}}
config2={"configurable": {"thread_id": "test2"}}
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

res = agent.invoke({"messages": [HumanMessage(content="Hi my name is Dev")]},config=config)
res
```

```python
res = agent.invoke({"messages": [HumanMessage(content="what is my name ?")]},config=config)
res 
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

res = agent.invoke({"messages": [HumanMessage(content="Hi my name is Hetvi")]},config=config2)
res
```
