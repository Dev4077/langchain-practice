Streaming and batching

Streaming

Most models can stream their output content while it is being generated. by Displaying output progressively, streming significanty improve user expreience, particularly for longer responcses. Calling stream() return an iterator that yield output chunk as they are produced. You can use a loop to process each chunk in real-time

```python
model.invoke("Give me a Joke on Samay Raina.")
```

```python
for chunk in model.stream("give me good youtube channel name for reseach based topic videos"):
    print(chunk.text, end="", flush=True)
```

Beatch

Batching a collection of independent req to a model can signifivantly improve performance and reduce costs, as the procesing can be done in parallel

```python
ress = model.batch([
    "why india is poor?",
    "What is moon expiry date?",
    "give me one short fuuny jokes in hindi"
], config={
    'max_concurrency' : 2, # Limit to paraller calls
})
for res in ress:
    print(res)
```
